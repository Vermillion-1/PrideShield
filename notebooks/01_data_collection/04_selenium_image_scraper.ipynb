```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time

# Set up Chrome options
chrome_options = Options()
chrome_options.add_argument("--headless")
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")
chrome_options.add_argument("--disable-gpu")
chrome_options.binary_location = "/usr/bin/google-chrome"
chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36")  # Set user-agent to avoid bot detection

# Set up ChromeDriver path
chrome_driver_path = "/usr/local/bin/chromedriver"
service = Service(chrome_driver_path)

# Create a WebDriver instance
driver = webdriver.Chrome(service=service, options=chrome_options)

# Open the Tumblr URL
# url = "https://unsplash.com/s/photos/lgbt"
# url = "https://imgflip.com/m/LGBTQ"
# url = "https://www.instagram.com/makeitgayer/?hl=en"
url = "https://www.reddit.com/r/lgbtmemes/top/"
driver.get(url)

# Wait for the page to load fully
time.sleep(10)  # Adjust if needed for slow loading pages

# Scroll to the bottom of the page to trigger lazy loading
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(5)  # Wait for images to load

# Find all image elements
img_tags = driver.find_elements(By.TAG_NAME, "img")

# Extract and print the image URLs
for img_tag in img_tags:
    # Check both 'src' and 'data-src' attributes for image URLs
    img_url = img_tag.get_attribute('src') or img_tag.get_attribute('data-src')
    if img_url:
        print(f"Image URL: {img_url}")

# Close the browser
driver.quit()
```

```python
import base64

# The base64 string (after removing the 'data:image/png;base64,' part)
base64_string = "<BASE64_IMAGE_REMOVED_FOR_CONTENT_SAFETY>"

# Decode the base64 string
image_data = base64.b64decode(base64_string)

# Write the decoded image to a file
with open("image.png", "wb") as file:
    file.write(image_data)

print("Image successfully saved as 'image.png'")
```
