# Label Reconciliation

Joins the two independent label files on row index and keeps only items where both labels agree,
producing the final training table.

**Effect on the dataset:** 1,667 labeled items → 1,320 retained (~21% dropped).

**Known limitation:** disagreements were dropped rather than adjudicated. Since annotators disagree
precisely on ambiguous content, this removes the hardest examples and inflates all downstream
metrics relative to real-world difficulty.
